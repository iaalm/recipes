```python
%pylab inline
g = 9.8
```

```python
def track(angle, c, v_init = 10):
    x, y = 0, 0
    ang_init = angle / 180 * pi
    dt = 1e-3

    tx = []
    ty = []
    vx, vy = v_init * cos(ang_init), v_init * sin(ang_init)
    print(vx, vy)
    while y >= 0:
        va = sqrt(vx * vx + vy * vy)
        ax, ay = - va * vx * c, - va * vy * c  # va ** 2 * (vx / va)
        ay -= g
        x, y = x + vx * dt + ax * dt * dt / 2, y + vy * dt + ay * dt * dt / 2
        vx, vy = vx + ax * dt, vy + ay * dt
        tx.append(x)
        ty.append(y)
    return tx, ty
```

```python
figure(figsize=(12,12))
for angle in (10, 15, 30, 45, 60):
    x, y = track(angle, 5)
    plot(x, y, label=angle)
legend()
show()
```

```python
def track_atmos(angle, c, v_init=10):
    x, y = 0, 0
    ang_init = angle / 180 * pi
    dt = 1e-4

    tx = []
    ty = []
    vx, vy = v_init * cos(ang_init), v_init * sin(ang_init)
    print(vx, vy)
    while y >= 0:
        va = sqrt(vx * vx + vy * vy)
        hc = c * exp(-y/4000)
        ax, ay = - va * vx * hc, - va * vy * hc
        ay -= g
        x, y = x + vx * dt + ax * dt * dt / 2, y + vy * dt + ay * dt * dt / 2
        vx, vy = vx + ax * dt, vy + ay * dt
        tx.append(x)
        ty.append(y)
    return tx, ty
```

```python
for angle in (20, 25, 30, 35, 40):
    x, y = track_atmos(angle, c = 0.1, v_init=5000)
    plot(x, y, label=angle)
legend()
show()
```
